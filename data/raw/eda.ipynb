```python
from pathlib import Path

results_dir = Path("./raw/results")
csv_files = sorted(results_dir.glob("*.csv"))

headers_by_sport = {}

for csv_file in csv_files:
    sport = csv_file.stem
    # nrows=0 reads only the header
    df = pd.read_csv(csv_file, nrows=0)
    headers_by_sport[sport] = df.columns.tolist()

all_headers = sorted(
    set().union(*headers_by_sport.values())
)

rows = []

for header in all_headers:
    sports = [
        sport
        for sport, headers in headers_by_sport.items()
        if header in headers
    ]

    if len(sports) == len(csv_files):
        coverage = "Common to everything"
    else:
        coverage = ", ".join(sorted(sports))

    rows.append({
        "header_field": header,
        "common_to_everything_or_sports": coverage,
        "count": len(sports)
    })

header_summary = pd.DataFrame(rows)

print(f"\nFound {len(all_headers)} unique header fields.\n")
display(header_summary.sort_values(by="count", ascending=False))
```

```python
from pathlib import Path
import pandas as pd

results_dir = Path("./raw/results")

all_result_types = set()

for csv_file in results_dir.glob("*.csv"):
    df = pd.read_csv(csv_file)

    if "result_type" in df.columns:
        all_result_types.update(
            df["result_type"].dropna().unique()
        )

for result_type in sorted(all_result_types):
    print(result_type)
```
